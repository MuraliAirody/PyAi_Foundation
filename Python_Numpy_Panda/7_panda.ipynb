```python
# Load the FIFA 2018 Statistics CSV file into a DataFrame.
# '../' moves one directory up from the current location.
df = pd.read_csv("../Datasets/datasets/FIFA 2018 Statistics.csv")

# ---------------------------------------------
# head() → Used to view the first few rows of a DataFrame.
#
# By default, df.head() returns the FIRST 5 rows.
# This is useful for quickly inspecting the structure,
# column names, and sample data without printing the entire table.
# ---------------------------------------------

print(df.head())      # Shows the first 5 rows (default)

print(df.head(10))    # Shows the first 10 rows (passing number explicitly)
```

# tail() → Displays the last few rows of the DataFrame.
# This is the opposite of head(), which shows the first rows.

# By default, df.tail() returns the LAST 5 rows.
# Useful when the dataset is large and you want to inspect
# how the data ends (e.g., check for missing values, formatting issues).

```python

print(df.tail())      # Shows the last 5 rows (default)

# You can also specify how many rows to view.
print(df.tail(10))    # Shows the last 10 rows
```
