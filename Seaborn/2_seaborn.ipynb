```python
import seaborn as sns   # Import seaborn for statistical data visualization

tips = sns.load_dataset("tips") # Load the 'tips' dataset from seaborn's built-in datasets

tips
```

```python

sns.relplot(x="total_bill", y="tip", data=tips, hue="smoker") 
#hue adds color based on smoker status (categorical variable)
```

```python
sns.relplot(x="total_bill", y="tip", data=tips, hue="smoker", style="time") 
#style changes marker based on time (Dinner/Lunch)
```

```python
sns.relplot(x="total_bill", y="tip", data=tips, hue="smoker", style="time", size='day') 
#size changes marker size based on day (categorical variable)
```

```python
sns.relplot(x="total_bill", y="tip", data=tips, hue="smoker", style="time", size='size') 
```

```python
sns.relplot(x="total_bill", y="tip", col='time', data=tips, hue="smoker", style="time", size='size')
 # col creates separate plots based on time (Dinner/Lunch)
```

```python
sns.catplot(x="day", y="total_bill", data=tips) 
# Basic categorical plot showing total bill by day
```

```python
sns.catplot(x="day", y="total_bill", data=tips, kind='swarm', hue='smoker')
# Swarm plot showing total bill by day, colored by smoker status
```

```python
sns.catplot(x="day", y="total_bill", data=tips, kind='swarm', hue='size')
# Swarm plot showing total bill by day, colored by size
```

```python
sns.catplot(x="day", y="total_bill", data=tips, kind='box', hue='sex')
# Box plot showing total bill by day, separated by sex
```
