```python
import seaborn as sns   # Import seaborn for statistical data visualization

# ----------------------------------------------------
# Get a list of built-in datasets available in seaborn
# These datasets are useful for practice and examples
# ----------------------------------------------------
sns.get_dataset_names()
```

```python

# ----------------------------------------------------
# Load the built-in "tips" dataset
# This dataset contains information about restaurant bills and tips
# head(10) limits the data to the first 10 rows for quick inspection
# ----------------------------------------------------
tips = sns.load_dataset("tips").head(10)

# ----------------------------------------------------
# relplot (relational plot)
# Used to visualize the relationship between two numerical variables
#
# x = "total_bill" → X-axis (total bill amount)
# y = "tip"        → Y-axis (tip given)
# data = tips      → DataFrame used for plotting
#
# By default, relplot creates a scatter plot
# ----------------------------------------------------
sns.relplot(x="total_bill", y="tip", data=tips)
```
