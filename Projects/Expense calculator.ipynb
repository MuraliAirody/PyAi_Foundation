# Expense Tracker (Menu-Driven Interface)

Success rate: 6.62%
Build a simple expense tracker that interacts via a menu-driven interface. Each expense has:
an auto-incremented integer ID (starting from 1)
a description (single word)
a positive floating-point amount
a category (single word)
When the program starts (and after each operation), display:
Add Expense
Delete Expense
Total Expenses
Category Total
List Expenses
Exit
Enter choice:
Then perform the selected operation:

#### Add Expense

- Print Description: and read the description.
- Print Amount: and read the amount.
- Print Category: and read the category.
- Record the new expense with the next ID.

#### Delete Expense
- Print Enter ID to delete: and read the ID.
- Remove the expense with that ID (if it exists).
#### Total Expenses
- Print the sum of all expenses (two decimal places).
####Category Total
- Print Enter category: and read the category.
- Print the sum of expenses in that category (two decimal places).
#### List Expenses
- Print every expense in insertion order, one per line:
<ID> <Description> <Amount> <Category>
#### Exit
- Terminate the program.

**Continue looping until the user selects 6.**


```python
print("1. Add Expense\n2. Delete Expense\n3. Total Expenses\n4. Category Total\n5. List Expenses\n6. Exit")

expenses = [[1,"house rent",10000.0,"rent"],
[2,"House rent",10000.0,"rent"],
[3,"Table rent",500.0,"Rent"]]

count = len(expenses)+1

while True:
    num = int(input("Enter choice: "))


    if num == 6:
        break
    elif num==1:
        desc = input("Description: ")
        amount = float(input("Amount: "))
        cat = input("Category: ")

        expense = list([count,desc,amount,cat])
        expenses.append(expense)

        count+=1
    elif num==2:
        expId = int(input("Print Enter ID to delete: ")) 
        for exp in expenses:
            if exp[0]==expId:
                expenses.remove(exp)
    elif num==3:
        total = 0
        for exp in expenses:
           total+=exp[2]
        print(f"{total:.2f}")   
    elif num==4:
        cat = input("Print Enter category: ")
        total = 0
        for exp in expenses:
            if (exp[3]).lower()==cat.lower():
                total+=exp[2]
        print(f"{total:.2f}")        
    elif num==5:
        for exp in expenses:
            print(f"{exp[0]} {exp[1]} {exp[2]} {exp[3]}")        

```
