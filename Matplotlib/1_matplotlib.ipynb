```python
import matplotlib.pyplot as plt   # Import pyplot module for creating plots

# X-axis data:
# Represents the independent variable (for example, day numbers)
week_days = [1, 2, 3, 4, 5, 6, 7, 8]

# Y-axis data:
# Represents the dependent variable (temperature values corresponding to each day)
temp = [21, 34, 56, 34, 42, 56, 40, 25]

# plot() draws a line graph:
# - week_days is plotted on the X-axis
# - temp is plotted on the Y-axis
# The values are connected in order using straight line segments
plt.plot(week_days, temp)
```

```python
# Plot temperature values against days
plt.plot(week_days, temp)

# Set label for the X-axis
# Describes what the values on the horizontal axis represent
plt.xlabel("Days")

# Set label for the Y-axis
# Describes what the values on the vertical axis represent
plt.ylabel("Temperature")

# Set the title of the graph
# Provides a short description of what the plot represents
plt.title("Weather Data")
```

```python
# Draw a line plot with customized color

# color="red" changes the line color to red
plt.plot(week_days, temp, color="red")

plt.xlabel("Days")
plt.ylabel("Temperature")
plt.title("Weather Data")
```

```python
# Draw a line plot with custom color and line style

# linestyle=":"   → draws a dotted line
# Supported line styles:
# '-', '--', '-.', ':', 'None', ' ', '', 'solid', 'dashed', 'dashdot', 'dotted'
plt.plot(week_days, temp, color="red", linestyle=":")

plt.xlabel("Days")
plt.ylabel("Temperature")
plt.title("Weather Data")
```

```python
# Data for X-axis: represents days (independent variable)
week_days = [1, 2, 3, 4, 5, 6, 7, 8]

# Y-axis data: temperature values for each day
# Maximum temperature recorded per day
temp_max = [21, 34, 56, 34, 42, 56, 40, 25]

# Minimum temperature recorded per day
temp_min = [19, 28, 40, 28, 35, 30, 32, 18]

# Average temperature calculated per day
temp_avg = [24, 30, 34, 30, 37, 37, 36, 24]

# Plot maximum temperature line
# label is used by legend() to identify this line
plt.plot(week_days, temp_max, label="Max Temperature")

# Plot minimum temperature line
plt.plot(week_days, temp_min, label="Min Temperature")

# Plot average temperature line
plt.plot(week_days, temp_avg, label="Avg Temperature")

# Display the legend on the plot
# Automatically maps labels to corresponding lines
plt.legend()

# Label the X-axis
plt.xlabel("Days")

# Label the Y-axis
plt.ylabel("Temperature")

# Set the title of the graph
# Describes the overall purpose of the visualization
plt.title("Daily Weather Temperature Comparison")
```

```python
# Bar Plot
# One column is categorical (Company names)
# The other column is numerical (Revenue values)

import matplotlib.pyplot as plt

# Categorical data for X-axis
# Each value represents a company
company = ["Amazon", "Google", "Tcs", "Wipro"]

# Numerical data for Y-axis
# Represents revenue corresponding to each company
revenue = [90, 60, 80, 100]

# Create a bar chart
# company → categories on X-axis
# revenue → height of each bar
# label   → used to display legend information
plt.bar(company, revenue, label="Revenue")

# Display legend on the plot
plt.legend()

# Label the X-axis to describe categories
plt.xlabel("Company")

# Label the Y-axis to describe values
plt.ylabel("Revenue")

# Set the title of the plot
# Explains what the bar chart represents
plt.title("Company Revenue Data")
```
