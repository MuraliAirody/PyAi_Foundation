## Column Explanations
- **transaction_id**: Unique ID for each transaction
- **customer_id**: ID mapping to individual customer
- **account_id**: Bank account from which transaction is made
- **transaction_date**: Date and time when transaction occurred
- **posting_date**: Date when transaction was posted to the bank
- **transaction_amount**: Amount debited or credited
- **currency**: Currency of the transaction
- **merchant_id**: Unique identifier for merchant
- **merchant_name**: Name of the business or vendor
- **merchant_category**: Type of goods/service purchased
- **card_issuer**: Issuer bank of the transaction card
- **location_city**: City where transaction took place
- **location_state**: State of transaction
- **location_country**: Country where merchant is located
- **balance_before**: Account balance before transaction
- **balance_after**: Account balance after transaction
- **is_fraud**: Boolean indicator if the transaction was fraudulent
- **customer_since**: Date the customer joined the bank
- **customer_age**: Age of the customer
- **customer_gender**: Gender identity of the customer
- **credit_score**: Credit score of customer
- **reward_points_earned**: Points earned from this transaction
