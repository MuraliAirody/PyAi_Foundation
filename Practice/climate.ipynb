# Average Rainfall 

```python
average_rainfall=master_data.groupby("City")["Rainfall (mm)"].transform("mean")

master_data["Average_Rainfall (mm)"] = average_rainfall

master_data[["City","State","Average_Rainfall (mm)"]]
```

# Compute monthly average temperature per city.

```python
master_data["Date"].dtype

master_data["Date"] = pd.to_datetime(master_data["Date"])


master_data["YearMonth"] = master_data["Date"].dt.to_period("M")

avg_temp_per_month = master_data.groupby(["City","YearMonth"])["Temperature_Avg (°C)"].mean().reset_index()

avg_temp_per_month = avg_temp_per_month.sort_values(["City","YearMonth"],ascending=[True,False])

avg_temp_per_month.groupby("City").head(3)
```

# Identify hottest and coldest day per city.

```python
id_hot = master_data.groupby("City")["Temperature_Avg (°C)"].idxmax()

master_data.loc[id_hot].sort_values("City")
```

```python
id_cold = master_data.groupby("City")["Temperature_Avg (°C)"].idxmin()

master_data.loc[id_cold].sort_values("City")
```

```python
hottest_days=master_data.loc[id_hot].sort_values("City")
coldest_days=master_data.loc[id_cold].sort_values("City")

extreme_days = pd.concat(
    [
        hottest_days.assign(Type="Hottest"),
        coldest_days.assign(Type="Coldest")
    ]
).sort_values(["City", "Type"])


extreme_days
```

# Find number of rainy days per city (Rainfall > 0).

```python
rain_fall = master_data[master_data["Rainfall (mm)"]>0].groupby("City").size()

rain_fall.reset_index(name="Rainfall days")
```

# Compare weekday vs weekend average AQI per city.

```python
master_data["Date"] = pd.to_datetime(master_data["Date"])

master_data["DateType"] = np.where(
    master_data["Date"].dt.weekday<5,"weekday","weekend"
)


daysGroupAQI = master_data.groupby(["City","DateType"])["AQI"].mean()

daysGroupAQI = daysGroupAQI.reset_index()


daysGroupAQI_Pivot = daysGroupAQI.pivot(index="City",columns="DateType",values="AQI").reset_index().round(2)

daysGroupAQI_Pivot
```

```python
master_data_grp = master_data[master_data["Rainfall (mm)"]!=0].groupby("City")["Rainfall (mm)"]


master_data_grp.mean().reset_index().round(2)
```
