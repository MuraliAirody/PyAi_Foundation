```python
name =" guru "

print(name.strip())
print(name.lstrip())
print(name.rstrip())
print(name.replace("u","U"))   
print(name.upper())
print(name.lower())
print(name.startswith(" g"))
print(name.endswith("u "))
print(name.split("r"))
print(name.find("r"))
print(name.index("u"))
print(name.count("r"))
print(len(name))
print(name.capitalize())
print(name.title())
print(name.center(20,"*"))
print(name.zfill(20))
print(name.swapcase())
print(name.isalpha())
print(name.isdigit())
print(name.isspace())
print(name.islower())
print(name.isupper())
print(name.isalnum())
print(name.partition("r"))
print(name.rpartition("r"))
print(name.encode())
print(name.expandtabs())
```

```python
str = "Hello World"

print(str.find("o")) # find first occurrence 
print(str.rfind("o")) # find last occurrence
print(str.index("o")) # index first occurrence
print(str.rindex("o")) # index last occurrence

# difference between find and index is find returns -1 if not found where as index raises an exception
print(str.find("z"))
try:
    print(str.index("z"))
except ValueError as e:
    print("ValueError:", e)

# find is safe to use when unsure if substring exists
```
